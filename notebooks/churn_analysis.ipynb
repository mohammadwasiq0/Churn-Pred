# Telecom Churn Prediction: Exploratory Data Analysis

In this notebook, we perform data analysis and visualization on the generated synthetic telecom churn dataset. This satisfies the requirement: `I want to use jupyter notebook for data analysis and visualization`.

## 1. Load Data

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import warnings
warnings.filterwarnings('ignore')

# Set aesthetic style
sns.set_style('whitegrid')

# Load data
data_path = '../data/churn_dataset.csv'
if os.path.exists(data_path):
    df = pd.read_csv(data_path)
    print(f"Data loaded successfully with shape: {df.shape}")
    display(df.head())
else:
    print("Dataset not found. Please run the data generator first.")
```

## 2. Check Class Balance

```python
if 'df' in locals():
    plt.figure(figsize=(8, 5))
    sns.countplot(x='Churn', data=df, palette='Set2')
    plt.title('Target Variable Distribution (Churn)')
    plt.show()
```

## 3. Correlation between Numerical Features

```python
if 'df' in locals():
    plt.figure(figsize=(10, 8))
    numerical_cols = ['tenure', 'MonthlyCharges', 'TotalCharges']
    
    # Handle TotalCharges missing/empty space issues
    df['TotalCharges'] = pd.to_numeric(df['TotalCharges'].replace(' ', None))
    
    corr_matrix = df[numerical_cols].corr()
    sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f')
    plt.title('Numerical Features Correlation Matrix')
    plt.show()
```

## 4. Churn by Contract Type

```python
if 'df' in locals():
    plt.figure(figsize=(10, 6))
    sns.countplot(x='Contract', hue='Churn', data=df, palette='pastel')
    plt.title('Churn Rate by Contract Type')
    plt.show()
```
